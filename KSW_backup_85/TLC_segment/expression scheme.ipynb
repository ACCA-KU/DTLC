```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
def plot(a1,b1,a3,b3,d1):
    x1 = np.linspace(0,0.5,50)
    x_eff = ((1-x1)-x1*d1)/((1-x1)+x1*(1-2*d1))
    x_ss = (1-x_eff)*a3/((1-x_eff)*a3+(x_eff)*b3)
    y1 = x_ss*a1 + (1-x_ss)*b1
    
    x2 = np.linspace(0.5,1,50)
    x_eff = (x2-(1-x2)*d1)/(x2+(1-x2)*(1-2*d1))
    x_ss = x_eff*a3/(x_eff*a3+(1-x_eff)*b3)
    y2 = x_ss*a1 + (1-x_ss)*b1
    return np.concatenate([x1,x2]), np.concatenate([y1,y2])
```

```python
x,y = plot(1,0,1,20,0.8)
plt.plot(x,y)
```
